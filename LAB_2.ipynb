```python
def group_and_sort(df, column):
    counts = df[column].value_counts().sort_values(ascending=False)
    return counts.reset_index().rename(columns={'index': column, column: 'count'})
```

```python
grouped = group_and_sort(df_sales, 'coffee_name')
print(grouped)
```
