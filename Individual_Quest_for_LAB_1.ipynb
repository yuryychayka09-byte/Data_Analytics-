Завдання 4: Масив із парних чисел

```python
arr = np.arange(2, 21, 2)   # перші 10 парних
print("Масив:", arr)
print("Сума:", arr.sum())
print("Середнє:", arr.mean())
print("Мінімум:", arr.min())
print("Максимум:", arr.max())
print("×3:", arr * 3)
```

```python
arr2d = np.random.randint(1, 11, (3,3))
print("Масив 3x3:\n", arr2d)
print("Сума по стовпцях:", arr2d.sum(axis=0))
print("Сума по рядках:", arr2d.sum(axis=1))
```
