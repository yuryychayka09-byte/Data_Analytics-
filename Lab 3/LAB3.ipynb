Step 4. Outlier Detection using IQR
Outliers were detected using the Interquartile Range (IQR) method. The analysis was performed on the value column because it contains the economic value of vehicles.

```python
# calculate IQR for value column

Q1 = df["value"].quantile(0.25)
Q3 = df["value"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(f"Q1 = {Q1}")
print(f"Q3 = {Q3}")
print(f"IQR = {IQR}")

print(f"Lower bound = {lower}")
print(f"Upper bound = {upper}")
```

```python
outliers = df[
    (df["value"] < lower) |
    (df["value"] > upper)
]

print("Number of outliers:", len(outliers))

outliers.head()
```

```python
df = df[
    (df["value"] >= lower) &
    (df["value"] <= upper)
]

print("Shape after outlier removal:")
print(df.shape)
```

Step 5----->

```python
df["release_date"] = pd.to_datetime(
    df["release_date"],
    errors="coerce"
)

print(df["release_date"].dtype)
```

Step 6. Data Type Conversion
The release_date column was converted from string format to datetime format.

```python
# convert release_date to datetime

df["release_date"] = pd.to_datetime(
    df["release_date"],
    errors="coerce"
)

print(df.dtypes)
```

Step 7. String Cleaning
Text columns were cleaned by removing extra spaces and converting values to lowercase.

```python
# clean text columns

df["country"] = (
    df["country"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["vehicle_type"] = (
    df["vehicle_type"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df["vehicle_sub_types"] = (
    df["vehicle_sub_types"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df[
    [
        "country",
        "vehicle_type",
        "vehicle_sub_types"
    ]
].head()
```

Step 8. Feature Engineering
New features were extracted from the release date column.

```python
# create new date features

df["year"] = df["release_date"].dt.year

df["month"] = df["release_date"].dt.month

df["day"] = df["release_date"].dt.day

df["day_of_week"] = df["release_date"].dt.dayofweek

df[
    [
        "release_date",
        "year",
        "month",
        "day",
        "day_of_week"
    ]
].head(10)
```

Step 9. One-Hot Encoding
The vehicle_type column was transformed into multiple binary columns using one-hot encoding.


```python
# one-hot encoding

df_encoded = pd.get_dummies(
    df,
    columns=["vehicle_type"]
)

print(df_encoded.head())
```

Step 10. Feature Scaling
The value column was normalized and standardized for further analysis.

```python
# min-max normalization

df["value_norm"] = (
    (df["value"] - df["value"].min()) /
    (df["value"].max() - df["value"].min())
)

# z-score standardization

df["value_std"] = (
    (df["value"] - df["value"].mean()) /
    df["value"].std()
)

print(
    df[
        [
            "value",
            "value_norm",
            "value_std"
        ]
    ].head()
)
```

```python
print("Min:", df["value_norm"].min())
print("Max:", df["value_norm"].max())

print("Mean:", df["value_std"].mean())
print("Std:", df["value_std"].std())
```

Step 11. Creating a New Feature
A binary feature was created using NumPy.

```python
df["is_expensive"] = np.where(
    df["value"] > df["value"].median(),
    1,
    0
)

print(
    df[
        [
            "value",
            "is_expensive"
        ]
    ].head(10)
)
```

Step 12. Reusable Cleaning Function
A reusable function was implemented to automate dataset cleaning.

```python
def clean_dataset(df):

    result = df.copy()

    # remove duplicates
    result = result.drop_duplicates()

    # fill missing numeric values
    numeric_cols = result.select_dtypes(
        include="number"
    ).columns

    for col in numeric_cols:
        result[col] = result[col].fillna(
            result[col].median()
        )

    # fill missing text values
    categorical_cols = result.select_dtypes(
        include="object"
    ).columns

    for col in categorical_cols:
        result[col] = result[col].fillna(
            result[col].mode()[0]
        )

    # convert date column
    if "release_date" in result.columns:

        result["release_date"] = pd.to_datetime(
            result["release_date"],
            errors="coerce"
        )

    return result
```

Step 13. Testing the Function
The cleaning function was tested on the vehicle dataset.

```python
clean_df = clean_dataset(df)

print("Original shape:", df.shape)
print("Cleaned shape:", clean_df.shape)

clean_df.head()
```

Step 14. Saving the Dataset
The cleaned dataset was exported to a CSV file.

```python
clean_df.to_csv(
    "general_vehicles_cleaned.csv",
    index=False
)

print("Dataset saved successfully.")
```
