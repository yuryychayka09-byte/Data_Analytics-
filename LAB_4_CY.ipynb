Step 1. Environment Setup and Data Loading

In this step, the necessary Python libraries are imported and the dataset containing information about War Thunder vehicles is loaded into the Google Colab environment.

The Pandas library is used for data manipulation and analysis, while NumPy provides additional numerical operations.

```python
import pandas as pd
import numpy as np

# Upload the dataset
from google.colab import files
uploaded = files.upload()

# Load the dataset
df = pd.read_csv('specific_vehicles.csv')

# Remove unnecessary index column
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Display the first five rows
df.head()
```

### Step 2. Grouping by One Column and Aggregation

In this step, the dataset is grouped by vehicle type. The total vehicle value is calculated for each category using the groupby() and sum() methods.


```python
# Calculate the total value for each vehicle type
value_by_type = df.groupby('vehicle_type')['value'].sum()

# Display the results
print(value_by_type)
```

The grouping operation produced a Series containing vehicle types and their total values.

This result allows us to compare the total value of different vehicle categories and identify which types have the highest overall value.

### Step 3. Calculating Average Values

In this step, the average vehicle value is calculated for each vehicle type. The groupby() method is used together with mean() to determine the average value within each category.

```python
# Calculate the average value for each vehicle type
avg_value = df.groupby('vehicle_type')['value'].mean()

# Display the results
print(avg_value)
```

The results show the average value of vehicles belonging to each category.

This information makes it possible to compare the average cost of different vehicle types and identify the most expensive categories.

### Step 4. Grouping by Multiple Columns

In this step, the dataset is grouped by two columns: country and vehicle_type.

The size() method is used to calculate the number of vehicles in each group.

```python
# Group vehicles by country and vehicle type
vehicle_count = df.groupby(['country', 'vehicle_type']).size()

# Display the results
print(vehicle_count)
```

```python
# Convert the grouped Series into a DataFrame
vehicle_count_df = df.groupby(
    ['country', 'vehicle_type']
).size().reset_index(name='count')

# Display the first ten records
print(vehicle_count_df.head(10))
```

The grouping operation produced a table containing countries, vehicle types and the number of corresponding records.

The reset_index() method converted the grouped result into a regular DataFrame, making it easier to process and visualize the information.

### Step 5. Using Multiple Aggregation Functions

In this step, the agg() method is used to calculate several statistical indicators simultaneously.

For each vehicle type, the total, average, minimum and maximum vehicle values are determined.

```python
# Calculate multiple statistical indicators
summary = df.groupby('vehicle_type')['value'].agg(
    ['sum', 'mean', 'min', 'max']
)

# Display the results
print(summary)
```

The resulting DataFrame contains four statistical indicators for each vehicle type.

The agg() method allows several calculations to be performed simultaneously, making the analysis more efficient.

### Step 6. Creating a Pivot Table

In this step, a pivot table is created to analyze the total vehicle value by country and vehicle type.

The pivot_table() method is used to organize the data into a structured table. Missing values are replaced with zeros.

```python
# Create a pivot table
pivot = pd.pivot_table(
    df,
    values='value',
    index='country',
    columns='vehicle_type',
    aggfunc='sum',
    fill_value=0
)

# Display the pivot table
print(pivot)
```

The pivot table displays the total value of each vehicle category for every country.

This representation makes it easier to compare vehicle categories across different countries and analyze the distribution of vehicle values.

### Step 7. Creating a Cross-Tabulation Table

In this step, the crosstab() method is used to calculate the frequency of vehicle types across different countries.

The resulting table shows how many vehicles of each category are represented for every country.

```python
# Create a cross-tabulation table
crosstab = pd.crosstab(
    df['country'],
    df['vehicle_type']
)

# Display the results
print(crosstab)
```

The cross-tabulation table shows the number of vehicles for each combination of country and vehicle type.

It provides an overview of the distribution of vehicle categories among different countries.

### Step 8. Merging DataFrames Using merge()

In this step, an additional DataFrame is created to classify vehicles into three general categories: Aircraft, Ground and Naval.

The merge() method is used to combine this information with the main dataset using the vehicle_type column.

```python
# Create a vehicle classification table
categories = pd.DataFrame({
    'vehicle_type': [
        'fighter',
        'bomber',
        'attack_helicopter',
        'utility_helicopter',
        'medium_tank',
        'light_tank',
        'heavy_tank',
        'tank_destroyer',
        'spaa',
        'assault',
        'boat',
        'heavy_boat',
        'destroyer',
        'light_cruiser',
        'heavy_cruiser',
        'battleship',
        'battlecruiser',
        'frigate',
        'barge'
    ],
    'category': [
        'Aircraft',
        'Aircraft',
        'Aircraft',
        'Aircraft',
        'Ground',
        'Ground',
        'Ground',
        'Ground',
        'Ground',
        'Ground',
        'Naval',
        'Naval',
        'Naval',
        'Naval',
        'Naval',
        'Naval',
        'Naval',
        'Naval',
        'Naval'
    ]
})

# Merge the classification table with the main dataset
df_merged = df.merge(
    categories,
    on='vehicle_type',
    how='left'
)

# Display the first five records
print(df_merged[['identifier', 'vehicle_type', 'category']].head())
```

The merge() operation successfully added a new category column to the main dataset.

Each vehicle is assigned to one of three general categories: Aircraft, Ground or Naval. The original records are preserved because a left join is used.

### Step 9. Combining DataFrames Using concat()

In this step, the original dataset is divided into two separate DataFrames based on the is_premium column.

The concat() method is then used to combine both parts into one DataFrame. The original indexes are preserved during the operation.

```python
# Split the dataset into two parts
df_part1 = df[df['is_premium'] == True]
df_part2 = df[df['is_premium'] == False]

# Combine both DataFrames
df_concat = pd.concat(
    [df_part1, df_part2],
    ignore_index=False
)

# Display the number of records
print(f"Part 1: {len(df_part1)}")
print(f"Part 2: {len(df_part2)}")
print(f"Combined dataset: {len(df_concat)}")

# Display the first five records
print(df_concat.head())
```

The two DataFrames were successfully combined into one dataset.

The total number of records in the combined DataFrame is equal to the number of records in the original dataset. The original indexes were preserved.

### Step 10. Combining DataFrames Using join()

In this step, an additional DataFrame is created with vehicle identifiers and repair costs.

The join() method is used to combine the main dataset with the additional information based on their indexes.

```python
# Create an additional DataFrame
repair_info = df[['repair_cost_arcade']].copy()

# Rename the column
repair_info = repair_info.rename(
    columns={'repair_cost_arcade': 'additional_repair_cost'}
)

# Combine DataFrames using their indexes
joined = df[['identifier', 'vehicle_type']].join(repair_info)

# Display the first five records
print(joined.head())
```

The join() method successfully combined the main information with the additional repair cost column.

The operation was performed using DataFrame indexes, allowing the vehicle identifiers, types and repair costs to be displayed in one table.

### Step 11. Calculating Analytical Indicators by Groups

In this step, the percentage contribution of each vehicle type to the total value of all vehicles is calculated.

First, the total value of the dataset is determined. Then, the total value of each vehicle category is divided by the overall value and multiplied by 100.

```python
# Calculate the total value of all vehicles
total_value = df['value'].sum()

# Calculate the percentage share for each vehicle type
value_share = (
    df.groupby('vehicle_type')['value'].sum()
    / total_value * 100
).round(2)

# Display the results
print(value_share)
```

The resulting Series shows the percentage contribution of each vehicle type to the total value of the dataset.

This indicator makes it possible to evaluate the distribution of vehicle values among different categories and identify the categories with the largest contribution.
